```python
class AioSchwab:
    _base_url = "https://api.schwabapi.com/trader/v1"

    def __init__(self):
        self._session = httpx.AsyncClient()

    async def __del__(self):
        await self._session.aclose()
    
    async def get(self, url: str, timeout: int = 10):
        url = f"{self._base_url}/{url}"
        response = await self._session.get(url, headers={"Authorization": f"Bearer {client.tokens.access_token}"}, timeout=timeout)
        return response
    
    async def account_numbers(self):
        response = await self.get("accounts/accountNumbers")
        return response.json()

aioSchwab = AioSchwab()
await aioSchwab.account_numbers()
```
